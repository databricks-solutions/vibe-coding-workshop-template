# Publish the workshop skills for Genie Code

Genie Code loads skills from your per-user skills folder, `/Workspace/Users/<your-email>/.assistant/skills/`. This notebook copies your workshop clone (`/Workspace/Users/<your-email>/vibe-coding-workshop`) into that folder, then checks that both are in place.

Use it when Genie Code can't do the copy for you. In its default **Auto-approve** mode, Genie Code always blocks its own writes to `.assistant/`, even after you confirm. Commands you run yourself in a notebook aren't checked.

**How to run:** attach this notebook to serverless compute and click **Run all**. The last cell should print two ✅ lines. Then go back to Genie Code and say the copy is done.

Safe to re-run: it replaces any previous copy.

```python
import os, subprocess

user = spark.sql("SELECT current_user()").first()[0]
env = {**os.environ, "D": f"/Workspace/Users/{user}"}
assert os.path.isdir(f"{env['D']}/vibe-coding-workshop/.git"), "Clone missing: run the git clone step first."

subprocess.run(["bash", "-c",
    'rm -rf "$D/.assistant/skills/vibe-coding-workshop"; mkdir -p "$D/.assistant/skills"; '
    'cp -R "$D/vibe-coding-workshop" "$D/.assistant/skills/"'], env=env, check=True, capture_output=True, text=True)
print(f"Copied {env['D']}/vibe-coding-workshop into {env['D']}/.assistant/skills/")
```

```python
result = subprocess.run(["bash", "-c",
    'test -d "$D/vibe-coding-workshop/.git" && echo "✅ 1/2 project cloned" || echo "❌ 1/2 re-run command 1"; '
    'test -f "$D/.assistant/skills/vibe-coding-workshop/skills/genie-code-environment/SKILL.md" && echo "✅ 2/2 skills published" || echo "❌ 2/2 re-run command 2"'],
    env=env, check=True, capture_output=True, text=True)
print(result.stdout)
```
